# Atributos de um Array NumPy
Os arrays NumPy possuem vários atributos que fornecem informações sobre sua estrutura e dados.

# dtype: Tipo de Dado dos Elementos
Indica o tipo de dado dos elementos armazenados no array (e.g., int64, float64).

```python
import numpy as np
```

```python
#Array int
array1 = np.array([5, 17, 32, -50])
print(array1)
print(array1.dtype)
```

```python
#Array float
array_float = np.array([1, 1.5, 3.4, -17, -3.2])
print(array_float)
print(array_float.dtype)
```

```python

#Array booleano
arr_bool = np.array([True, False, True])
print(arr_bool)
print(arr_bool.dtype)
```

```python
#Array string
array_str = np.array(['Python', 'Java', 'C++'])
print(array_str)
print(array_str.dtype)
```

# shape: Formato do Array
Retorna uma tupla indicando o número de elementos em cada dimensão.

```python
array_1dim = np.array([10, 43, 22])
print(array_1dim)
print(array_1dim.shape)
```

```python
matriz = np.array([[10, 49, 37], [-13, -9, -15]])
print(matriz)
print(matriz.shape)
```

# size: Número Total de Elementos
Retorna o número total de elementos no array.

```python
matriz2 = np.array([[10, 49, 37], [-13, -9, -15]])
print(matriz2.size)
```

```python
arr_1d = np.array([1, 2, 3, 4, 5])
print("Size de arr_1d:", arr_1d.size)
```

# ndim: Número de Dimensões
Retorna o número de dimensões (eixos) do array. É o mesmo que a dimensão do array, conforme vimos anteriormente.

```python
array_1d = np.array([2.3, 2.5, 9.7])
print(array_1d.ndim)
```

```python
matriz = np.array([[10, 49, 37], [-13, -9, -15]])
print(matriz.ndim)
```

# Conversões de Arrays
Você pode converter o tipo de dado de um array usando o método .astype().

```python
array_float = np.array([1, 1.5, 3.4, -17, -3.2])
print(array_float)
print(array_float.dtype)

#cast
array_cast = array_float.astype(int)
print(array_cast)
print(array_cast.dtype)
```

```python
# Convertendo para float
arr3 = np.array([1, 2, 3])

print(arr3)
print(arr3.dtype)

arr4 = arr3.astype(float)
print(arr4)
print(arr4.dtype)
```
