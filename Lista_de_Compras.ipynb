```python
itens = []

def main():
    while True:
        print("\nMenu:")
        print("1. Adicionar item")
        print("2. Listar itens")
        print("3. Marcar item como comprado")
        print("4. Remover item")
        print("5. Sair")

        escolha = input("Escolha uma opção: ")

        if escolha == "1":
            nome_item = input("Digite o nome do item: ")
            adicionar_item(nome_item)
        elif escolha == "2":
            listar_item()
        elif escolha == "3":
          listar_item()
          try:
            indice = int(input('Digite o índice do item que deseja marcar como comprado: '))
            marcar_item_comprado(indice - 1)
            listar_item()
          except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")
        elif escolha == "4":
          listar_item()
          try:
            indice = int(input('Digite o índice do item que quer remover: '))
            remover_item(indice - 1)
            listar_item()
          except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")
        elif escolha == "5":
          print('Sair')
          break
        else:
           print('opção inválida!')


def adicionar_item(nome_item):
    itens.append({'descrição': nome_item, 'Comprado' : False})
    print(f"Item '{nome_item}' adiciona com sucesso!")

def listar_item():
  if not itens:
    print('Não existe nenhum item')
    return
  print('\nLista de Itens:')
  for i, item in enumerate(itens):
    status = '[X]' if item['Comprado'] else '[ ]'
    print(f"{i + 1}. {status} {item['descrição']}")


def marcar_item_comprado(indice):
  if 0 <= indice <= len(itens):
    itens[indice]['Comprado'] = True
    print(f"Item '{itens[indice]['descrição']}' marcado como comprado!")
  else:
    print('Índice inválido!')


def remover_item(indice):
  if 0 <= indice <= len(itens):
    item_removido = itens.pop(indice)
    print(f"Item '{item_removido['descrição']}' removido com sucesso!")

```

```python
main()
```
