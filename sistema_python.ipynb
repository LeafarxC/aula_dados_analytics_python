```python
tarefas = []

def main():
  print("Bem vindo")
  while True:
    print("\nMenu:")
    print("1. Adicionar Tarefa")
    print("2. Ver tarefa")
    print("3. Concluir tarefa")
    print("4. Remover tarefa")
    print("5. Sair")

    escolha = input("Escolha uma opção: ")

    if escolha == "1":
      descricao = input("Digite a descrição da tarefa: ")
      adicionar_tarefas(descricao)
    elif escolha == "2":
      ver_tarefas()
    elif escolha == "3":
      ver_tarefas()
      try:
        indice = int(input("Digite o índice da tarefa a ser concluída: ")) - 1
        concluir_tarefa(indice)
      except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")
    elif escolha == '4':
      ver_tarefas()
      try:
        indice = int(input("Digite o índice da tarefa a ser removida: ")) - 1
        remover_tarefa(indice)
      except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")
    elif escolha == '5':
      print('Sair')
      break
    else:
      print('opção inválida!')



def adicionar_tarefas(descricao):
  tarefas.append({'descrição': descricao, 'concluida' : False})
  print(f"Tarefa '{descricao}' adiciona com sucesso!")


def ver_tarefas():
  if not tarefas:
    print('Não existe nenhuma tarefa')
    return
  print('\nLista de Tarefas:')
  for i, tarefa in enumerate(tarefas):
    status = '[X]' if tarefa['concluida'] else '[ ]'
    print(f"{i + 1}. {status} {tarefa['descrição']}")


def concluir_tarefa(indice):
  if 0 <= indice <= len(tarefas):
    tarefas[indice]['concluida'] = True
    print(f"Tarefa '{tarefas[indice]['descrição']}' concluída!")
  else:
    print("Índice inválido")

def remover_tarefa(indice):
  if 0 <= indice <= len(tarefas):
    tarefa_removida = tarefas.pop(indice)
    print(f"Tarefa '{tarefa_removida['descrição']}' removida!")
```

```python
main()
```
