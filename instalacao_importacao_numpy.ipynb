```python
#Importação do Numpy
import numpy as np
```

```python
#Diferença no tempo de execução entre uma lista e um array numpy
from time import time
inicio = time()
lista1 = list(range(10**8))
soma1 = 0

for i in lista1:
    soma1 += i

fim = time()
tempo1 = fim - inicio
print(f'Tempo de execução da lista: {tempo1}')
```

```python
#Diferença no tempo de execução entre uma lista e um array numpy
inicio2 = time()
arr = np.array(lista1, dtype=float)
soma2 = arr.sum()

fim2 = time()
tempo2 = fim2 - inicio2
print(tempo2)
```

```python
print(tempo2 / tempo1)
```
