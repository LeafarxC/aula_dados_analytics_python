# 📊 O que são Quartis
Imagine uma fila de 100 pessoas ordenadas por altura. Se você dividir essa fila em 4 partes iguais, cada parte terá 25 pessoas.

Quartis dividem os dados ordenados em quatro partes iguais:

Q1 (1º quartil): 25% dos dados estão abaixo

*   Q1 (1º quartil): 25% dos dados estão abaixo
*   Q2 (Mediana): 50% dos dados estão abaixo
*   Q3 (3º quartil): 75% dos dados estão abaixo


Objetivo:

✅ Entender a distribuição dos dados

✅ Detectar outliers (valores extremos)

✅ Segmentar dados em faixas (ex: clientes de baixo/médio/alto valor)

✅ Criar gráficos tipo boxplot

✅ Fazer análises comparativas com mais contexto!

```python
import pandas as pd

# Carregar base de dados
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'
df = pd.read_csv(url)

# Exibir primeiras linhas
df.head()
```

# 📐 Calculando os Quartis
Vamos aplicar a função quantile() no campo total_bill.

```python
quartis = df['total_bill'].quantile([0.25, 0.5, 0.75])
print("Quartis:")
print(quartis)
```

```python
df['total_bill'].median()
```

```python
df['total_bill'].quantile(0.10)  # 10% dos conjuntos de dados
```

```python
quartis = df['tip'].quantile([0.25, 0.5, 0.75])
print("Quartis:")
print(quartis)
```

```python
df['total_bill'].mean()
```

```python
df['total_bill'].describe()
```

# 📦 Classificando por Quartis
Você também pode usar qcut() para categorizar os dados em faixas baseadas nos quartis.

```python
df['segmento'] = pd.qcut(df['total_bill'], q=4, labels=['baixo', 'médio-baixo', 'médio-alto', 'alto'])
df[['total_bill', 'segmento']].head()
```
