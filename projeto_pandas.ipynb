```python
#calculos (pandas e numpy) descobrir a receita total

print(f"Rceita total de vendas da loja: {df_vendas['Receita_total'].sum():.2f}") # .2f - 2 casas decimais
```

```python
#preço médio unitário dos produtos
print(f"Preço médio dos produtos: {df_vendas['Preco_Unitario'].mean():.2f}")
```

```python
#quantidade minima e máxima de produtos vendidos
print(f"Quantidade mínima de produtos vendidos: {df_vendas['Quantidade'].min()}")
print(f"Quantidade máxima de produtos vendidos: {df_vendas['Quantidade'].max()}")
```

```python
#simular valores ausentes dentro do df
df_vendas_com_nan = df_vendas.copy() #novo dataframe, foi feito uma copia
df_vendas_com_nan.loc[df_vendas_com_nan.sample(frac=0.05).index, 'Preco_Unitario'] = np.nan
df_vendas_com_nan.loc[df_vendas_com_nan.sample(frac=0.03).index, 'Quantidade'] = np.nan

```

```python
df_vendas_com_nan.head()
```

```python
df_vendas_com_nan.info()
```

```python
df_vendas_com_nan.isna().sum()
```

```python
#Preencher com a media os valores NAN da coluna quantidade
df_vendas_pr = df_vendas_com_nan.copy()
media_qtd = df_vendas_pr['Quantidade'].mean()
df_vendas_pr['Quantidade'].fillna(media_qtd, inplace=True)
```

```python
df_vendas_pr.info()
```

```python
#remover os valores ausentes
df_vendas_lp = df_vendas_pr.dropna(subset=['Preco_Unitario'])
```

```python
df_vendas_lp.info()
```

```python
df_vendas_lp.head()
```

```python
vendas_por_produto = df_vendas_lp.groupby('Produto').agg(
    Quantidade_Total=('Quantidade', 'sum'),
    Receita_Total=('Receita_total', 'sum'),
).reset_index()
```

```python
print(vendas_por_produto)
```

```python
#OPrdenar os valores por quantidade total para saber a quantidade mais vendida
produto_mais_vendido = vendas_por_produto.sort_values(by='Quantidade_Total', ascending=False).iloc[0]
print(produto_mais_vendido)
```

```python
#Ordenar os valores por receita total para saber o maior valor total gerado
produto_maior_receita = vendas_por_produto.sort_values(by='Receita_Total', ascending=False).iloc[0]
print(produto_maior_receita)
```

```python
df_produtos = pd.read_csv('/content/Produtos.csv')
```

```python
df_produtos.head()
```

```python
#converter a coluna Data venda para tipo correto data
df_vendas_lp['Data_Venda'] = pd.to_datetime(df_vendas_lp['Data_Venda'])
```

```python
df_vendas_lp.info()
```

```python
#analisar as vendas por mês
df_vendas_lp['Mes_Venda'] = df_vendas_lp['Data_Venda'].dt.month
```

```python
df_vendas_lp.head()
```

```python
venda_por_mes = df_vendas_lp.groupby('Mes_Venda')['Receita_total'].sum().reset_index()
```

```python
print(venda_por_mes)
```

```python
print(df_produtos)
```

```python
df_merge = pd.merge(df_vendas_lp, df_produtos, on='Produto', how='left')
```

```python
print(df_merge)
```

```python
#Analise de vendas por categoria
vendas_por_categoria = df_merge.groupby('Categoria')['Receita_total'].sum().reset_index()
```

```python
print(vendas_por_categoria)
```

```python
#filtros dentro do DataFrame
df_merge.head()
```

```python
#filtrar por coluna ( Produto, quantidade, receita  total)
df_colunas_filtradas = df_merge.filter(items=['Produto', 'Quantidade', 'Receita_total'])
```

```python
print(df_colunas_filtradas)
```

```python
#filtragem mais avançada
df_query = df_merge.query('Produto == "Teclado" and Quantidade > 10')
```
